## 3. Variables: labelled storage (15 min)

A **variable** gives a meaningful name to a value. The single equals sign `=` means **assign**: put the value on the right into the name on the left. It does not mean mathematical equality.

Use names that convey quantity and units: `flow_rate_m3_h` is safer than `x`.

```python
flow_rate_m3_h = 12.5
operating_hours = 10
volume_m3 = flow_rate_m3_h * operating_hours

print('Processed volume:', volume_m3, 'm³')
```

**Think:** If you change `flow_rate_m3_h`, will `volume_m3` update by itself? Run the calculation cell again to find out.

### Good variable names
Start with a letter or `_`; use letters, digits, and `_`; do not use spaces. Python is case-sensitive: `Temperature_C` and `temperature_c` differ.

```python
inlet_temperature_c = 50
outlet_temperature_c = 100
temperature_rise_c = outlet_temperature_c - inlet_temperature_c
print('Temperature rise =', temperature_rise_c, '°C')

# inlet temperature c = 25  # Invalid: spaces are not allowed
```

## 5. Mini-practice: tank hold-up (10 min)

A tank has volume 3.0 m³ and a liquid flow rate of 0.50 m³/min. Estimate residence time: $t = V/Q$.

1. Write variables with units in their names.
2. Calculate `residence_time_min`.
3. Print a readable answer.

```python
tank_volume_m3 = 3.0
flow_rate_m3_min = 0.50

residence_time_min = tank_volume_m3 / flow_rate_m3_min

print(f'Residence time = {residence_time_min:.2f} min')

# Change the flow rate and rerun to see the new residence time
# flow_rate_m3_min = 0.25
# residence_time_min = tank_volume_m3 / flow_rate_m3_min
# print(f'Residence time = {residence_time_min:.2f} min')
```
