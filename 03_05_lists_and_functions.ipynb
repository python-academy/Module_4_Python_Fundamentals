## 1. Lists: one name, many related values (15 min)

A **list** stores an ordered collection inside square brackets. Lists are ideal for a short series of readings. Positions, called **indices**, begin at 0: the first reading is `readings[0]`.

```python
flow_rates_m3_h = [10.2, 10.5, 9.9, 10.1, 10.4]

print('All readings:', flow_rates_m3_h)
print('First reading:', flow_rates_m3_h[0], 'm³/h')
print('Last reading:', flow_rates_m3_h[-1], 'm³/h')
print('Number of readings:', len(flow_rates_m3_h))
```

## 2. Changing and summarising lists (12 min)

Lists are **mutable**: their contents can change. Common methods include `.append()` to add an item. Built-in functions such as `sum()`, `min()`, and `max()` help create quick summaries. `sum(values) / len(values)` is the arithmetic mean.

```python
temperatures_c = [68.5, 70.1, 69.8]
temperatures_c.append(71.0)
average_c = sum(temperatures_c) / len(temperatures_c)

print('Readings:', temperatures_c)
print(f'Average = {average_c:.2f} °C')
print('Range:', min(temperatures_c), 'to', max(temperatures_c), '°C')
```

**Try it (5 min):** Add one value to your pH list and calculate its average.

## 3. Functions: reusable calculation recipes (15 min)

A **function** is a named recipe. `def` defines it; inputs are **parameters**; `return` sends a result back. Functions reduce repetition and make code easier to test. Define the function cell before calling it.

```python
def residence_time(volume_m3, flow_rate_m3_h):
    """Return ideal residence time in hours."""
    return volume_m3 / flow_rate_m3_h

tau_h = residence_time(12.0, 3.0)
print(f'Residence time = {tau_h:.2f} h')
```

**Try it (5 min):** Call `residence_time` for two other tanks. What happens if the flow rate is 0? Why?

**Challenge:** Add a condition in the function that returns a clear message or `None` when flow rate is not positive.

```python
def residence_time(volume_m3, flow_rate_m3_h):
    """Return ideal residence time in hours, or None for a nonpositive flow."""
    if flow_rate_m3_h <= 0:
        return None
    return volume_m3 / flow_rate_m3_h


tau_h1 = residence_time(18.0, 3.0)
tau_h2 = residence_time(55.0, 5.0)
print(f'Tank 1 residence time = {tau_h1:.2f} h')
print(f'Tank 2 residence time = {tau_h2:.2f} h')

tau_h3 = residence_time(12.0, 0.0)
if tau_h3 is None:
    print("Residence time unavailable: flow rate must be greater than zero.")
```

## 4. Functions used with a list (10 min)

This pattern applies one carefully tested calculation to every value in a list.

```python
def celsius_to_kelvin(temperature_c):
    return temperature_c + 273.15

measurements_c = [20, 35.5, 80]
for measurement_c in measurements_c:
    print(f'{measurement_c} °C = {celsius_to_kelvin(measurement_c):.2f} K')
```
