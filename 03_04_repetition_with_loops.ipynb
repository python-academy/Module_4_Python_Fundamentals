Use loops to apply the same calculation or check to multiple readings. Automation is often simply a reliable repeatable task applied to a collection of data.

## 1. Why loops? (8 min)

If we need to inspect ten temperature readings, copying an `if` statement ten times is slow and error-prone. A loop says: *for each item, carry out these steps.*

We will use `for` loops first because they are the safest starting point when the number of repetitions is known.

```python
for minute in range(1, 6):
    print(f'Reading at minute {minute}')
```

`range(1, 6)` creates 1, 2, 3, 4, 5: the final number is excluded.

## 2. Repeated calculations (15 min)

Indent all instructions that should repeat. Here, the same conversion calculation is applied at several times.

```python
for time_min in range(0, 31, 5):
    conversion_percent = 100 * (1 - 0.9**time_min)
    print(f'{time_min:2d} min → conversion = {conversion_percent:6.2f}%')
```

**Try it (5 min):** Print the mass processed at 1, 2, 3, 4, and 5 hours for a flow rate of 250 kg/h.

**Challenge:** Calculate a running total using `total = total + value` inside the loop. What must be initialized before the loop?

```python
flow_rate = 250  # kg/h
total = 0        # initialize before the loop

for hour in range(1, 6):
    value = flow_rate
    total = total + value
    print(f"After {hour} hour(s): {total} kg")
```

## 3. `while` loops: repeat until a condition changes (12 min)

A `while` loop repeats as long as its condition remains true. Make sure something inside the loop changes; otherwise the loop can run forever. For early learning, prefer `for` when possible.

```python
tank_level_percent = 100

while tank_level_percent > 20:
    print(f'Current level: {tank_level_percent}%')
    tank_level_percent = tank_level_percent - 20

print('Low-level threshold reached.')
```

**Think:** What line prevents an infinite loop? Change the initial level to 65 and inspect the result.

**Challenge:** Use a `while` loop to find how many equal 10% reductions are needed to take a concentration from 100 to below 50 mg/L.

```python
concentration = 100.0

while concentration > 50:
    print(f'Current concentration: {concentration:.1f} mg/L')
    concentration = concentration * 0.9
```

## 4. Combining loops and decisions (15 min)

A useful pattern is: loop through readings, then use a conditional to label each one. The code below is a simple alarm scan—not a real safety system.

```python
temperatures_c = [72, 84, 96, 88, 101]

for temperature_c in temperatures_c:
    if temperature_c > 100:
        label = 'ALARM'
    elif temperature_c > 85:
        label = 'CAUTION'
    else:
        label = 'normal'
    print(f'{temperature_c} °C → {label}')
```

**Pair task (10 min):** A storage vessel must stay between 30% and 85% full. Loop through `[25, 40, 90, 70, 30]` and label each reading `LOW`, `OK`, or `HIGH`.

**Challenge:** Count how many readings are outside the range and print one final summary.

```python
levels_percent = [25, 40, 90, 70, 30]
outside_count = 0

for level in levels_percent:
    if level < 30:
        label = "LOW"
        outside_count += 1
    elif level > 85:
        label = "HIGH"
        outside_count += 1
    else:
        label = "OK"

    print(f"{level}% -> {label}")

print(f"Readings outside the range: {outside_count}")
```
